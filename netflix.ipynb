```python
import numpy as np
import pandas as pd 
import matplotlib.pyplot as plt
import seaborn as sns
```

```python
netflix=pd.read_csv('netflix_titles.csv')
netflix
```

```python
netflix.shape
```

```python
netflix.head()
```

```python
netflix.describe()
```

```python
netflix.info()
```

```python
netflix.isnull().sum()
```

```python
netflix.fillna({
    'director': 'Unknown',
    'cast': 'Unknown',
    'country': 'Unknown'
}, inplace=True)

```

```python
netflix.dropna(subset=['date_added', 'rating', 'duration'], inplace=True)
```

```python
netflix.isnull().sum()
```

```python
netflix.duplicated().sum()
```

```python
netflix['date_added'] = pd.to_datetime(netflix['date_added'].str.strip())
```

```python
netflix.dtypes
```

```python
netflix['year_added']=netflix['date_added'].dt.year
netflix['month_added']=netflix['date_added'].dt.month
```

```python
netflix.sample(2)
```

```python
netflix[['date_added', 'year_added', 'month_added']].head()
```

```python
netflix[['type', 'duration']].head(10)
```

```python
netflix['duration_int'] = netflix['duration'].str.extract(r'(\d+)').astype(float)
```

```python
netflix[['type', 'duration', 'duration_int']].head(10)
```

```python
netflix.groupby('type')['duration_int'].mean()
```

```python
netflix.groupby('type')['duration_int'].agg(['mean', 'min', 'max', 'count'])
```

```python
netflix['type'].value_counts()
```

```python

netflix['type'].value_counts().plot(kind='bar', color=['#E50914', '#221f1f'])
plt.title('Movies vs TV Shows on Netflix')
plt.xlabel('Type')
plt.ylabel('Count')
plt.show()
```

```python
netflix['country'].str.split(', ').explode().value_counts()
```

```python
netflix['country'].str.split(', ').explode().value_counts().head(10).plot(kind='barh')
plt.title('Top 10 Countries by Content Count')
plt.xlabel('Count')
plt.show()
```

```python
top_genres = netflix['listed_in'].str.split(', ').explode().value_counts().head(10)
print(top_genres)
```

```python
top_genres.plot(kind='barh', color='#E50914')
plt.title('Top 10 Genres on Netflix')
plt.xlabel('Count')
plt.gca().invert_yaxis()  
plt.show()
```

```python
yearly_content = netflix['year_added'].value_counts().sort_index()
print(yearly_content)
```

```python
netflix['year_added'].value_counts().sort_index().plot(kind='line')   # ✅
```

```python
netflix['year_added'].value_counts().sort_index().plot(kind='line', marker='o', color='#E50914')
plt.title('Netflix Content Added Per Year')
plt.xlabel('Year')
plt.ylabel('Number of Titles Added')
plt.grid(True)
plt.show()
```

```python
netflix.groupby(['year_added', 'type']).size().unstack().plot(kind='line', marker='o')
plt.title('Movies vs TV Shows Growth Over Years')
plt.show()
```

```python
netflix['rating'].value_counts()
```

```python

plt.figure(figsize=(10,6))
sns.countplot(data=netflix, x='rating', order=netflix['rating'].value_counts().index, color='#E50914')
plt.title('Content Rating Distribution on Netflix')
plt.xlabel('Rating')
plt.ylabel('Count')
plt.xticks(rotation=45)
plt.show()
```

```python
netflix['rating'].unique()
```

```python
top_directors = netflix[netflix['director'] != 'Unknown']['director'].str.split(', ').explode().value_counts().head(10)
print(top_directors)
```

```python
top_actors = netflix[netflix['cast'] != 'Unknown']['cast'].str.split(', ').explode().value_counts().head(10)
print(top_actors)
```

```python
top_directors.plot(kind='barh', color='#E50914')
plt.title('Top 10 Directors on Netflix')
plt.gca().invert_yaxis()
plt.show()
```

```python
top_actors.plot(kind='barh', color='#E50914')
plt.title('Top 10 Directors on Netflix')
plt.gca().invert_yaxis()
plt.show()
```

```python
movies = netflix[netflix['type'] == 'Movie']
movies
```

```python
plt.figure(figsize=(10,6))
plt.hist(movies['duration_int'], bins=30, color='#E50914', edgecolor='black')
plt.title('Distribution of Movie Durations on Netflix')
plt.xlabel('Duration (minutes)')
plt.ylabel('Number of Movies')
plt.show()
```
